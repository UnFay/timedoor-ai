```python
import pandas as pd
students = [['Tom', 90], ['Jack', 80], ['Jill', 70]]
data_students = pd.DataFrame(students, columns=['Name', 'Grade'])

new_students = [['Sam', 60], ['Mary', 50], ['John', 40], ['Joseph', 30]]
new_students = pd.DataFrame(new_students, columns=['Name', 'Grade'])
data_students = pd.concat([data_students, new_students], ignore_index=True)
data_students.to_csv('data_students.csv')
```
